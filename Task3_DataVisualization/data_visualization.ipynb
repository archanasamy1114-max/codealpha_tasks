```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("dataset/netflix_titles.csv")
```

```python
df.head()
```

```python
df.head()
```

```python
df["type"].value_counts()
```

```python
df["type"].value_counts().plot(kind="bar")

plt.title("Movies vs TV Shows")
plt.xlabel("Type")
plt.ylabel("Number of Titles")

plt.show()
```

```python
df["rating"].value_counts().head(10)
```

```python
df["rating"].value_counts().head(10).plot(kind="bar")

plt.title("Top 10 Netflix Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Titles")

plt.show()
```

```python
year_counts = df["release_year"].value_counts().sort_index()

year_counts.plot(figsize=(12,5))

plt.title("Netflix Titles by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.show()
```

```python
df["country"].value_counts().head(10)
```

```python
df["country"].value_counts().head(10).plot(kind="bar", figsize=(10,5))

plt.title("Top 10 Countries by Netflix Titles")
plt.xlabel("Country")
plt.ylabel("Number of Titles")

plt.xticks(rotation=45)

plt.show()
```

```python
df["listed_in"].value_counts().head(10)
```

```python
df["listed_in"].value_counts().head(10).plot(kind="bar", figsize=(10,5))

plt.title("Top 10 Netflix Genres")
plt.xlabel("Genre")
plt.ylabel("Number of Titles")

plt.xticks(rotation=45)

plt.show()
```

```python
df["type"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(6,6)
)

plt.title("Movies vs TV Shows Distribution")
plt.ylabel("")

plt.show()
```

# Data Visualization Summary

The visualizations show important patterns in the Netflix dataset.

- Movies are more numerous than TV Shows.
- TV-MA and TV-14 are among the most common ratings.
- Netflix titles increased significantly in recent years.
- The United States has the highest number of titles in the country analysis.
- Different genres contribute significantly to the Netflix content library.

## Key Findings

1. Movies form a larger portion of the Netflix dataset than TV Shows.
2. TV-MA is one of the most common content ratings.
3. Netflix content shows strong growth in recent release years.
4. The United States contributes the highest number of titles in the country analysis.
5. The dataset contains a wide variety of genres available on Netflix.
