## Key Findings


1. The dataset contains a total of 8,807 Netflix titles, including 6,131 Movies and 2,676 TV Shows. Movies account for a larger share of the dataset than TV Shows.

2. The most common content rating is TV-MA, with 3,207 titles, followed by TV-14 with 2,160 titles. This shows that mature-rated content forms a significant portion of the dataset.

3. The United States has the highest number of titles in the country field, with 2,818 titles, followed by India with 972 titles. This indicates that the United States and India are major contributors to the Netflix content dataset.

4. Netflix content increased significantly in recent years, with 2018 having the highest number of titles by release year in the dataset. This indicates strong growth in content production and availability during the recent period.

5. The dataset contains missing values in several columns, especially director, cast and country. These missing values were identified and handled during the data cleaning process to improve the quality of the analysis.
